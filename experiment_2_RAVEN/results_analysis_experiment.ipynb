## Check what is the accuracy in each group

```python
def extract_question_num(s):
    last_parts = s.split('_')[-2:]
    if last_parts[-1].isdigit():
        result = last_parts[-1]
    else:
        result = "_".join(last_parts)
    return result
```

```python
def convert_2_int(num):
    if pd.isna(num) or num == '-':
        return None
    return int(num)

def pred_for_group_A(question):
    questions = {
        '219': 6,
        '1228_a': 1,
        '1229': 7,
        '6569': 1,
        '7108': 8,
        '128': 2,
        '218': 5,
        '429': 7,
        '1228_b': 2,
        '6999': 1
    }
    return questions[question]

def analysis_test_results(users_logs, group_id):
    results = {}

    users_logs = users_logs[users_logs["ugroup"] == group_id]
    users = users_logs['tuid'].unique()
    
    for user in users:
        user_data = users_logs[users_logs["tuid"] == user]

        correct_count = 0
        correct_pred_count = 0 # for questions that pred == true
        incorrect_pred_count = 0 # for questions that pred != true
        
        questions = 0
        total_confidence = 0
        wrong_pred_questions = 0 # total pred!=true questions
        right_pred_questions = 0 # toal pred=true questions

        answer_pred_user = 0 # total questions where user_choice = pred
                
        for _, row in user_data.iterrows():
            question = extract_question_num(row['blockId'])
            true_choice = convert_2_int(row['trueShapeChoiceIndex'])
            pred_choice = convert_2_int(row['predictedShapeChoiceIndex']) if group_id!='A' else pred_for_group_A(question)
            user_choice = convert_2_int(row['selectedShapeIndex'])
            confidence = convert_2_int(row['faceSelected'])
            
            if pd.notna(row['stepActionType']):
                questions += 1
                total_confidence += confidence
                
                # check how many qeusiotns does the user answered correctly
                if user_choice == true_choice:
                    correct_count += 1
    
                    if true_choice == pred_choice:
                        correct_pred_count += 1
                        
                    else:
                        incorrect_pred_count += 1
    
                # count how many correct and incorrect qeustions there are in the data
                if true_choice == pred_choice:
                    right_pred_questions += 1
    
                else:
                    wrong_pred_questions += 1

                # count how many questions the user and the pred where right
                if pred_choice == user_choice:
                    answer_pred_user += 1


        results[user] = {
            "correct": correct_count,
            "questions": questions,
            "confidence": total_confidence,
            "correct_pred_count": correct_pred_count,
            "incorrect_pred_count": incorrect_pred_count,
            "right_pred_questions": right_pred_questions,
            "wrong_pred_questions": wrong_pred_questions,
            "answer_pred_user": answer_pred_user
        }

    return results
```

```python
# create a syntetic group - F
def is_fully_confident(question_id):
    return question_id in [
        '218', 
        '219',
        '1228_b',
        '1229',
        '6999',
        '7108'
    ]
    
def create_group_F(users_logs):
    results = {}

    users_logs = users_logs[users_logs["ugroup"] == 'E']
    users = users_logs['tuid'].unique()
    
    for user in users:
        user_data = users_logs[users_logs["tuid"] == user]

        correct_count = 0
        correct_pred_count = 0 # for questions that pred == true
        incorrect_pred_count = 0 # for questions that pred != true
        
        questions = 0
        total_confidence = 0
        wrong_pred_questions = 0 # total pred!=true questions
        right_pred_questions = 0 # toal pred=true questions

        answer_pred_user = 0 # total questions where user_choice = pred
                
        for _, row in user_data.iterrows():
            question = extract_question_num(row['blockId'])
            true_choice = convert_2_int(row['trueShapeChoiceIndex'])
            pred_choice = convert_2_int(row['predictedShapeChoiceIndex'])
            question = extract_question_num(row['blockId'])

            is_fully_confident_question = is_fully_confident(question)
            confidence = 5 if is_fully_confident_question else convert_2_int(row['faceSelected'])
            user_choice = pred_choice if is_fully_confident_question else convert_2_int(row['selectedShapeIndex'])
            
                
            if pd.notna(row['stepActionType']):
                questions += 1
                total_confidence += confidence
                
                # check how many qeusiotns does the user answered correctly
                if user_choice == true_choice:
                    correct_count += 1
    
                    if true_choice == pred_choice:
                        correct_pred_count += 1
                        
                    else:
                        incorrect_pred_count += 1
    
                # count how many correct and incorrect qeustions there are in the data
                if true_choice == pred_choice:
                    right_pred_questions += 1
    
                else:
                    wrong_pred_questions += 1

                # count how many questions the user and the pred where right
                if pred_choice == user_choice:
                    answer_pred_user += 1


        results[user] = {
            "correct": correct_count,
            "questions": questions,
            "confidence": total_confidence,
            "correct_pred_count": correct_pred_count,
            "incorrect_pred_count": incorrect_pred_count,
            "right_pred_questions": right_pred_questions,
            "wrong_pred_questions": wrong_pred_questions,
            "answer_pred_user": answer_pred_user
        }

    return results
```

```python
group_a_data = analysis_test_results(df, "A")
group_b_data = analysis_test_results(df, "B")
group_c_data = analysis_test_results(df, "C")
group_d_data = analysis_test_results(df, "D")
group_e_data = analysis_test_results(df, "E")
group_f_data = create_group_F(df)
```

```python
def get_success_per_group(group_data):
    correct = 0
    total_questions = 0

    correct_pred_count = 0
    incorrect_pred_count = 0

    right_pred_questions = 0
    wrong_pred_qeustions = 0

    answer_pred_user = 0
    
    for user in group_data:
        correct += group_data[user]['correct']
        total_questions += group_data[user]['questions']
        # total_questions += group_data[user]['right_pred_questions']+group_data[user]['wrong_pred_questions']

        if 'correct_pred_count' and 'incorrect_pred_count' in group_data[user]:
            correct_pred_count += group_data[user]['correct_pred_count']
            incorrect_pred_count += group_data[user]['incorrect_pred_count']

        if 'right_pred_questions' and 'wrong_pred_questions' in group_data[user]:
            right_pred_questions += group_data[user]['right_pred_questions']
            wrong_pred_qeustions += group_data[user]['wrong_pred_questions']

        if 'answer_pred_user' in group_data[user]:
            answer_pred_user += group_data[user]['answer_pred_user']

    # accuracy = f"{(correct / total_questions) * 100:.1f}"
    return correct, total_questions, len(group_data), correct_pred_count, incorrect_pred_count, right_pred_questions, wrong_pred_qeustions, answer_pred_user
```

# Check what is the confidence rate

```python
def calculate_mean_confidence(group_data, group_name=""):
    total_confidence = 0
    total_questions = 0

    for i, user in enumerate(group_data, start=1):
        user_conf = group_data[user]['confidence']
        user_q = group_data[user]['questions']

        total_confidence += user_conf
        total_questions += user_q

        user_mean = user_conf / user_q if user_q else 0
        # print(f"User #{i}: mean confidence = {user_mean:.1f}")

    overall_mean = total_confidence / total_questions if total_questions else 0
    print(f"Group {group_name} overall mean confidence: {overall_mean:.3f}")
    return overall_mean
```

## Check the post hoc questions results

```python
def mean_last_3_answers(users_logs, group_id):
    if group_id == "C":
        explanation = "LLM"
    elif group_id == "D":
        explanation = "OS"
    elif group_id == "E":
        explanation = ''
    
    users_data = users_logs[users_logs['ugroup'] == group_id]

    users_num = users_data['tuid'].unique()
    
    question1_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_1']
    question2_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_2']
    question3_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_3']
    
    q1_sum = question1_rows['faceSelected'].sum()
    q2_sum = question2_rows['faceSelected'].sum()
    q3_sum = question3_rows['faceSelected'].sum()

    return q1_sum, q2_sum, q3_sum, len(users_num)
```

```python
*llm_means, llm_participants = mean_last_3_answers(df, "C")
*os_means, os_participants = mean_last_3_answers(df, "D")
*prediction_means, prediction_participants = mean_last_3_answers(df, "E")

questions = ['Q1: Understanding', 'Q2: Clarity', 'Q3: Trust']
x = range(len(questions))
width = 0.25  # Reduce width to fit 3 bars per group

fig, ax = plt.subplots()

# Adjust x positions for three bars
ax.bar([i - width for i in x], [v / llm_participants for v in llm_means], width, label='LLM (Group C)')
ax.bar(x, [v / os_participants for v in os_means], width, label='OS (Group D)')
ax.bar([i + width for i in x], [v / prediction_participants for v in prediction_means], width, label='Model Confidence (Group E)')

ax.set_ylabel('Average Rating')
ax.set_title('Average Ratings per Question by Explanation Type')
ax.set_xticks(x)
ax.set_xticklabels(questions)
ax.set_ylim(0, 5)
ax.legend()

plt.tight_layout()
plt.show()
```

```python
def group_2_num(g):
    if g == 'A':
        return 1
    elif g == 'B':
        return 2
    elif g == 'C':
        return 3
    elif g == 'D':
        return 4
    elif g == 'E':
        return 5
    return 6
    
def test_results(users_logs, group_id):
    results = {}

    users_logs = users_logs[users_logs["ugroup"] == (group_id if group_id!='F' else 'E')]
    users = users_logs['tuid'].unique()
    
    for user in users:
        user_data = users_logs[users_logs["tuid"] == user]
        user_group = group_2_num(group_id)
        
        for _, row in user_data.iterrows():
            true_choice = convert_2_int(row['trueShapeChoiceIndex'])
            pred_choice = convert_2_int(row['predictedShapeChoiceIndex'])
            user_choice = convert_2_int(row['selectedShapeIndex'])
            confidence = convert_2_int(row['faceSelected'])
            question = extract_question_num(row['blockId'])
            is_correct = 0 

            if group_id == 'F':
                is_fully_confident_question = is_fully_confident(question)
                confidence = 5 if is_fully_confident_question else convert_2_int(row['faceSelected'])
                user_choice = pred_choice if is_fully_confident_question else convert_2_int(row['selectedShapeIndex'])
            
            # check if these are the only question rows
            if pd.notna(row['stepActionType']):
                
                # check how many qeusiotns does the user answered correctly
                if user_choice == true_choice:
                    is_correct = 1


                question_details = {
                    'Group': user_group,
                    'Question': question,
                    'UserChoice': user_choice,
                    # 'True Choice': true_choice,
                    'IsAnswerCorrect': int(is_correct),
                    'Confidence': confidence
                    
                }

                user_key = f"F_{user}" if group_id == "F" else user
                
                if user_key not in results:
                    results[user_key] = []
                
                results[user_key].append(question_details)


    return results
```

```python
def get_confidence(q):
    confidence = {
        '128': 34.5,
        '218': 33.9,
        '6999': 90.3,
        '7108': 46.7,
        '219': 44.2,
        '429': 26.5,
        '1228_a': 28.5,
        '1228_b': 66,
        '1229': 79.4,
        '6569': 41.7,
    }
    return confidence[q]

def extract_question_num(s):
    last_parts = s.split('_')[-2:]
    if last_parts[-1].isdigit():
        result = last_parts[-1]
    else:
        result = "_".join(last_parts)
    return result
```

```python
def is_invalid_user(user):
    invalid_users = [
        'UID_Pv48eG17sn92Ge75sg',
        'UID_ZV94Gd16sF84aL62eH',
        'UID_TEST_TEST_1',
        'UID_UA64iV84bd20uJ55aO',
        'UID_Kg24ef54aE73wd38JF',
        'UID_pk73wE89io45VZ91BF',
        'UID_JG54wN91du61Ls34Ng',
        'UID_UA64iV84bd20uJ55aO',
        'UID_TEST_TEST_1',
        'UID_ZV94Gd16sF84aL62eH',
        'UID_Pv48eG17sn92Ge75sg'
    ]
    
    return user in invalid_users
```

```python
def check_pred_vs_user_per_question(users_logs, group_id):
    questions = {}

    is_F = group_id == "F"
    group_id = group_id if group_id != "F" else "E"
    users_logs = users_logs[users_logs["ugroup"] == group_id]
    users = users_logs['tuid'].unique()
    
    for user in users:
        if is_invalid_user(user):
            continue
            
        user_data = users_logs[users_logs["tuid"] == user]
                
        for _, row in user_data.iterrows():
            question = extract_question_num(row['blockId'])
            true_choice = convert_2_int(row['trueShapeChoiceIndex'])
            pred_choice = convert_2_int(row['predictedShapeChoiceIndex'])
            user_choice = convert_2_int(row['selectedShapeIndex'])

            if pd.notna(row['stepActionType']):
                if question not in questions:
                    questions[question] = {
                        'confidence': get_confidence(question),
                        'matched_answers': 0,
                        'questions_appearance': 0
                    }
    
                questions[question]['questions_appearance'] += 1

                if is_F:
                    if is_fully_confident(question) or user_choice == pred_choice:
                        questions[question]['matched_answers'] += 1
                    
                elif user_choice == pred_choice:
                    questions[question]['matched_answers'] += 1

    return questions
```

```python
def mean_last_3_answers(users_logs, group_id):
    if group_id == "C":
        explanation = "LLM"
    elif group_id == "D":
        explanation = "OS"
    elif group_id == "E":
        explanation = ''

    results = {
        "q_1": {i: 0 for i in range(1, 6)},
        "q_2": {i: 0 for i in range(1, 6)},
        "q_3": {i: 0 for i in range(1, 6)},
    }
    
    users_data = users_logs[
        (users_logs['ugroup'] == group_id) &
        (~users_logs['tuid'].apply(is_invalid_user))
    ]

    users_num = users_data['tuid'].unique()
    
    question1_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_1']
    question2_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_2']
    question3_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_3']

    for _, row in question1_rows.iterrows():
        confidence = convert_2_int(row['faceSelected'])
        results['q_1'][confidence] += 1

    for _, row in question2_rows.iterrows():
        confidence = convert_2_int(row['faceSelected'])
        results['q_2'][confidence] += 1
        
    for _, row in question3_rows.iterrows():
        confidence = convert_2_int(row['faceSelected'])
        results['q_3'][confidence] += 1
        
    return results
```
