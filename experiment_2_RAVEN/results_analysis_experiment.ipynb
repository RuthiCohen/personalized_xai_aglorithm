## Check what is the accuracy in each group

```python
def extract_question_num(s):
    last_parts = s.split('_')[-2:]
    if last_parts[-1].isdigit():
        result = last_parts[-1]
    else:
        result = "_".join(last_parts)
    return result
```

```python
def convert_2_int(num):
    if pd.isna(num) or num == '-':
        return None
    return int(num)

def pred_for_group_A(question):
    questions = {
        '219': 6,
        '1228_a': 1,
        '1229': 7,
        '6569': 1,
        '7108': 8,
        '128': 2,
        '218': 5,
        '429': 7,
        '1228_b': 2,
        '6999': 1
    }
    return questions[question]

def analysis_test_results(users_logs, group_id):
    results = {}

    users_logs = users_logs[users_logs["ugroup"] == group_id]
    users = users_logs['tuid'].unique()
    
    for user in users:
        user_data = users_logs[users_logs["tuid"] == user]

        correct_count = 0
        correct_pred_count = 0 # for questions that pred == true
        incorrect_pred_count = 0 # for questions that pred != true
        
        questions = 0
        total_confidence = 0
        wrong_pred_questions = 0 # total pred!=true questions
        right_pred_questions = 0 # toal pred=true questions

        answer_pred_user = 0 # total questions where user_choice = pred
                
        for _, row in user_data.iterrows():
            question = extract_question_num(row['blockId'])
            true_choice = convert_2_int(row['trueShapeChoiceIndex'])
            pred_choice = convert_2_int(row['predictedShapeChoiceIndex']) if group_id!='A' else pred_for_group_A(question)
            user_choice = convert_2_int(row['selectedShapeIndex'])
            confidence = convert_2_int(row['faceSelected'])
            
            if pd.notna(row['stepActionType']):
                questions += 1
                total_confidence += confidence
                
                # check how many qeusiotns does the user answered correctly
                if user_choice == true_choice:
                    correct_count += 1
    
                    if true_choice == pred_choice:
                        correct_pred_count += 1
                        
                    else:
                        incorrect_pred_count += 1
    
                # count how many correct and incorrect qeustions there are in the data
                if true_choice == pred_choice:
                    right_pred_questions += 1
    
                else:
                    wrong_pred_questions += 1

                # count how many questions the user and the pred where right
                if pred_choice == user_choice:
                    answer_pred_user += 1


        results[user] = {
            "correct": correct_count,
            "questions": questions,
            "confidence": total_confidence,
            "correct_pred_count": correct_pred_count,
            "incorrect_pred_count": incorrect_pred_count,
            "right_pred_questions": right_pred_questions,
            "wrong_pred_questions": wrong_pred_questions,
            "answer_pred_user": answer_pred_user
        }

    return results
```

```python
# create a syntetic group - F
def is_fully_confident(question_id):
    return question_id in [
        '218', 
        '219',
        '1228_b',
        '1229',
        '6999',
        '7108'
    ]
    
def create_group_F(users_logs):
    results = {}

    users_logs = users_logs[users_logs["ugroup"] == 'E']
    users = users_logs['tuid'].unique()
    
    for user in users:
        user_data = users_logs[users_logs["tuid"] == user]

        correct_count = 0
        correct_pred_count = 0 # for questions that pred == true
        incorrect_pred_count = 0 # for questions that pred != true
        
        questions = 0
        total_confidence = 0
        wrong_pred_questions = 0 # total pred!=true questions
        right_pred_questions = 0 # toal pred=true questions

        answer_pred_user = 0 # total questions where user_choice = pred
                
        for _, row in user_data.iterrows():
            question = extract_question_num(row['blockId'])
            true_choice = convert_2_int(row['trueShapeChoiceIndex'])
            pred_choice = convert_2_int(row['predictedShapeChoiceIndex'])
            question = extract_question_num(row['blockId'])

            is_fully_confident_question = is_fully_confident(question)
            confidence = 5 if is_fully_confident_question else convert_2_int(row['faceSelected'])
            user_choice = pred_choice if is_fully_confident_question else convert_2_int(row['selectedShapeIndex'])
            
                
            if pd.notna(row['stepActionType']):
                questions += 1
                total_confidence += confidence
                
                # check how many qeusiotns does the user answered correctly
                if user_choice == true_choice:
                    correct_count += 1
    
                    if true_choice == pred_choice:
                        correct_pred_count += 1
                        
                    else:
                        incorrect_pred_count += 1
    
                # count how many correct and incorrect qeustions there are in the data
                if true_choice == pred_choice:
                    right_pred_questions += 1
    
                else:
                    wrong_pred_questions += 1

                # count how many questions the user and the pred where right
                if pred_choice == user_choice:
                    answer_pred_user += 1


        results[user] = {
            "correct": correct_count,
            "questions": questions,
            "confidence": total_confidence,
            "correct_pred_count": correct_pred_count,
            "incorrect_pred_count": incorrect_pred_count,
            "right_pred_questions": right_pred_questions,
            "wrong_pred_questions": wrong_pred_questions,
            "answer_pred_user": answer_pred_user
        }

    return results
```

```python
group_a_data = analysis_test_results(df, "A")
group_b_data = analysis_test_results(df, "B")
group_c_data = analysis_test_results(df, "C")
group_d_data = analysis_test_results(df, "D")
group_e_data = analysis_test_results(df, "E")
group_f_data = create_group_F(df)
```

```python
def get_success_per_group(group_data):
    correct = 0
    total_questions = 0

    correct_pred_count = 0
    incorrect_pred_count = 0

    right_pred_questions = 0
    wrong_pred_qeustions = 0

    answer_pred_user = 0
    
    for user in group_data:
        correct += group_data[user]['correct']
        total_questions += group_data[user]['questions']
        # total_questions += group_data[user]['right_pred_questions']+group_data[user]['wrong_pred_questions']

        if 'correct_pred_count' and 'incorrect_pred_count' in group_data[user]:
            correct_pred_count += group_data[user]['correct_pred_count']
            incorrect_pred_count += group_data[user]['incorrect_pred_count']

        if 'right_pred_questions' and 'wrong_pred_questions' in group_data[user]:
            right_pred_questions += group_data[user]['right_pred_questions']
            wrong_pred_qeustions += group_data[user]['wrong_pred_questions']

        if 'answer_pred_user' in group_data[user]:
            answer_pred_user += group_data[user]['answer_pred_user']

    # accuracy = f"{(correct / total_questions) * 100:.1f}"
    return correct, total_questions, len(group_data), correct_pred_count, incorrect_pred_count, right_pred_questions, wrong_pred_qeustions, answer_pred_user
```

```python
print('------------ Accuracy Table: User Choice vs. True Choice -------------')
rows = []
for group_id, data in groups.items():

    correct, total, participants, correct_pred, incorrect_pred, right_pred_questions, wrong_pred_quesionts, answer_pred_user = get_success_per_group(data)
    accuracy = round((correct / total) * 100, 1)

    row = {
        'Group': group_id,
        'Correct answers': correct,
        'Total questions': total,
        'Accuracy': f"{accuracy}%",
        'Participants': participants, 
        'Correctly Predicted Answers': f"{correct_pred}/{right_pred_questions} ~ {(correct_pred/right_pred_questions)*100:.1f}%",
        'Incorrectly Predicted Answers': f"{incorrect_pred}/{wrong_pred_quesionts} ~ {(incorrect_pred/wrong_pred_quesionts)*100:.1f}%",
    }
    rows.append(row)

df_result = pd.DataFrame(rows)
df_result
```

## graph: plot 2: A scatter plot for "Error Detection Gap" 
X-Axis: Agreement Rate - Percentage of trials where the human followed the AI when the AI was correct.

Y-Axis: Recovery Rate - Percentage of trials where the human corrected the AI when the AI was incorrect.

Divide the plot into four quadrants using a vertical and horizontal dashed line at the 50\% mark.

Top-Right: "High Synergy" (High reliance + High recovery).

Bottom-Right: "Persuasion Paradox" (High reliance + Low recovery).

Use different colored symbols for groups, circle, triangle, box, etc.

```python
import re
import matplotlib.pyplot as plt

# ---------------- Plot: Agreement vs Recovery (Quadrants) ----------------
# X = Agreement Rate (% trials human followed AI when AI was correct)
# Y = Recovery Rate  (% trials human corrected AI when AI was incorrect)

def _extract_percent(s: str) -> float:
    """
    Extract the percentage number from strings like:
    '91/120 ~ 75.8%'  -> 75.8
    """
    if s is None:
        return float('nan')
    m = re.search(r'~\s*([0-9]+(?:\.[0-9]+)?)\s*%', str(s))
    return float(m.group(1)) if m else float('nan')

# Compute per-group points from df_result
df_plot = df_result.copy()
df_plot['AgreementRate'] = df_plot['Correctly Predicted Answers'].apply(_extract_percent)
df_plot['RecoveryRate']  = df_plot['Incorrectly Predicted Answers'].apply(_extract_percent)

# Consistent colors (reuse your Plot 2 color map)
group_colors = {
    'A': '#1f77b4',  # blue
    'B': '#ff7f0e',  # orange
    'C': '#2ca02c',  # green
    'D': '#d62728',  # red
    'E': '#9467bd',  # purple
    'F': '#8c564b',  # brown
}

# Different symbols per group (circle, triangle, box, etc.)
group_markers = {
    'A': 'o',  # circle
    'B': '^',  # triangle up
    'C': 's',  # square
    'D': 'D',  # diamond
    'E': 'P',  # plus-filled
    'F': 'X',  # X-filled
}

fig, ax = plt.subplots(figsize=(8.5, 6))

# Scatter points (one per group)
for _, row in df_plot.iterrows():
    g = row['Group']
    x = row['AgreementRate']
    y = row['RecoveryRate']
    ax.scatter(
        x, y,
        s=120,
        marker=group_markers.get(g, 'o'),
        color=group_colors.get(g, None),
        edgecolors='black',
        linewidths=0.8,
        label=f"Group {g}"
    )
    # Label each point with the group letter
    ax.text(
        x + 1.5,
        y - 1.5,
        f"({x:.1f}%, {y:.1f}%)",
        fontsize=9,
        ha='left',
        va='top'
    )

# Axes
ax.set_xlabel('Agreement Rate (%)\n(Human followed AI when AI was correct)')
ax.set_ylabel('Recovery Rate (%)\n(Human corrected AI when AI was incorrect)')
ax.set_xlim(0, 100)
ax.set_ylim(0, 100)

# Quadrant lines at 50%
ax.axvline(50, linestyle='--', linewidth=1)
ax.axhline(50, linestyle='--', linewidth=1)

# Quadrant labels
ax.text(75, 95, 'High Synergy', ha='center', va='top', fontsize=10)
ax.text(75, 5,  'Persuasion Paradox', ha='center', va='bottom', fontsize=10)

# (Optional) label other quadrants (not required, but can help)
# ax.text(25, 95, 'Low Reliance + High Recovery', ha='center', va='top', fontsize=9)
# ax.text(25, 5,  'Low Reliance + Low Recovery',  ha='center', va='bottom', fontsize=9)

ax.grid(True, linestyle=':', linewidth=0.8)
ax.set_title('Agreement vs. Recovery by Group', fontsize=12)

# Legend (avoid duplicates)
handles, labels = ax.get_legend_handles_labels()
by_label = dict(zip(labels, handles))
ax.legend(by_label.values(), by_label.keys(), loc='lower left', fontsize=9, frameon=True)

plt.tight_layout()

# ---- Save image (PNG + PDF) ----
plt.savefig("study2_error_detection_gap.png", dpi=300, bbox_inches="tight")
plt.show()

# Quick check: see the computed values
df_plot[['Group', 'AgreementRate', 'RecoveryRate']]
```

## graph: plot 3: Subjective Rating Radar Chart
Axes: Use three axes radiating from a center point: Trust, Clarity, and Understanding.

Shape: Each group should be represented by a closed polygon (a triangle in this case).

Transparency: Use a semi-transparent fill so the reader can see where the shapes overlap.

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------- Data ----------------
labels = ['Understanding', 'Clarity', 'Trust']

data = {
    'C': [4.1, 3.5, 3.95],
    'D': [3.5238095238095237, 3.4285714285714284, 3.6666666666666665],
    'E': [4.208333333333333, 4.416666666666667, 4.541666666666667]
}

# Consistent colors (same scheme as previous plots)
group_colors = {
    'A': '#1f77b4',
    'B': '#ff7f0e',
    'C': '#2ca02c',  # green
    'D': '#d62728',  # red
    'E': '#9467bd',  # purple
    'F': '#8c564b',
}

# ---------------- Radar setup ----------------
num_vars = len(labels)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]  # close the loop

# ---- Layout: radar + table ----
fig = plt.figure(figsize=(8, 8))
gs = fig.add_gridspec(2, 1, height_ratios=[4, 1.2])

ax = fig.add_subplot(gs[0], polar=True)

# Axis labels
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
ax.set_thetagrids(np.degrees(angles[:-1]), labels, fontsize=10)

# Radial scale (Likert 1–5)
ax.set_ylim(1, 5)
ax.set_yticks([1, 2, 3, 4, 5])
ax.set_yticklabels(['1', '2', '3', '4', '5'], fontsize=9)

# ---------------- Plot each group ----------------
for group, values in data.items():
    values_closed = values + values[:1]  # close polygon
    ax.plot(
        angles,
        values_closed,
        linewidth=2,
        label=f'Group {group}',
        color=group_colors[group]
    )
    ax.fill(
        angles,
        values_closed,
        color=group_colors[group],
        alpha=0.25
    )

# ---------------- Radar title & legend ----------------
ax.set_title('Perceived Trust, Clarity, and Understanding by Group', fontsize=12, pad=15)
ax.legend(loc='upper right', bbox_to_anchor=(1.25, 1.1), fontsize=9)

# ---------------- Table (separate axis) ----------------
ax_tbl = fig.add_subplot(gs[1])
ax_tbl.axis('off')

row_labels = [f'Group {g}' for g in data.keys()]
col_labels = labels
cell_text = [[f"{v:.2f}" for v in data[g]] for g in data.keys()]

table = ax_tbl.table(
    cellText=cell_text,
    rowLabels=row_labels,
    colLabels=col_labels,
    cellLoc='center',
    rowLoc='center',
    loc='center'
)

table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 1.4)

# Color row labels to match group colors
for i, g in enumerate(data.keys(), start=1):  # row 0 is header
    table[(i, -1)].get_text().set_color(group_colors[g])
    table[(i, -1)].get_text().set_weight('bold')

plt.tight_layout()
plt.savefig("study2_subjective_rating_radar_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

# Check what is the confidence rate

```python
def calculate_mean_confidence(group_data, group_name=""):
    total_confidence = 0
    total_questions = 0

    for i, user in enumerate(group_data, start=1):
        user_conf = group_data[user]['confidence']
        user_q = group_data[user]['questions']

        total_confidence += user_conf
        total_questions += user_q

        user_mean = user_conf / user_q if user_q else 0
        # print(f"User #{i}: mean confidence = {user_mean:.1f}")

    overall_mean = total_confidence / total_questions if total_questions else 0
    print(f"Group {group_name} overall mean confidence: {overall_mean:.3f}")
    return overall_mean
```

```python
def mean_last_3_answers(users_logs, group_id):
    if group_id == "C":
        explanation = "LLM"
    elif group_id == "D":
        explanation = "OS"
    elif group_id == "E":
        explanation = ''
    
    users_data = users_logs[users_logs['ugroup'] == group_id]

    users_num = users_data['tuid'].unique()
    
    question1_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_1']
    question2_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_2']
    question3_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_3']
    
    q1_sum = question1_rows['faceSelected'].sum()
    q2_sum = question2_rows['faceSelected'].sum()
    q3_sum = question3_rows['faceSelected'].sum()

    return q1_sum, q2_sum, q3_sum, len(users_num)
```

```python
import numpy as np
from collections import Counter

def print_sample_demographics(groups, users_details):
    participant_ids = []

    for group in groups:
        participant_ids.extend(group.keys())

    # Prevent accidental duplication
    participant_ids = list(dict.fromkeys(participant_ids))

    ages = np.array(
        [
            float(users_details[user]["age"])
            for user in participant_ids
            if user in users_details
        ]
    )

    genders = [
        str(users_details[user]["gender"]).lower()
        for user in participant_ids
        if user in users_details
    ]

    print("Participants:", len(participant_ids))
    print("Participants with age data:", len(ages))
    print(f"Mean age: {ages.mean():.2f}")
    print(f"Sample SD: {ages.std(ddof=1):.2f}")
    print(f"Age range: {ages.min():.0f}–{ages.max():.0f}")
    print("Gender counts:", Counter(genders))
```

```python
def mean_last_3_answers(users_logs, group_id):
    if group_id == "C":
        explanation = "LLM"
    elif group_id == "D":
        explanation = "OS"
    elif group_id == "E":
        explanation = ''

    results = {
        "q_1": {i: 0 for i in range(1, 6)},
        "q_2": {i: 0 for i in range(1, 6)},
        "q_3": {i: 0 for i in range(1, 6)},
    }
    
    users_data = users_logs[
        (users_logs['ugroup'] == group_id) &
        (~users_logs['tuid'].apply(is_invalid_user))
    ]

    users_num = users_data['tuid'].unique()
    
    question1_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_1']
    question2_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_2']
    question3_rows = users_data[users_data['blockId'] == f'block_end_questionnaire_explanation_{explanation}_question_3']

    for _, row in question1_rows.iterrows():
        confidence = convert_2_int(row['faceSelected'])
        results['q_1'][confidence] += 1

    for _, row in question2_rows.iterrows():
        confidence = convert_2_int(row['faceSelected'])
        results['q_2'][confidence] += 1
        
    for _, row in question3_rows.iterrows():
        confidence = convert_2_int(row['faceSelected'])
        results['q_3'][confidence] += 1
        
    return results
```
